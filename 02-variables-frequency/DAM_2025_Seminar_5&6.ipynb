Another way to make subplots is to use `plt.subplots` - it cerates an array with areas for plots, that may be axesed using indexing

For example, if we want 2, rows, 3 columns we use `fig, axes=plt.subplots(nrows=2,ncols=3)`

image.png

This method allows working with subplots as with 1D/2D arrays, BUT requires changes:



*   Instead of `plt` need to use `axes[...]` (`plt.plot`, `plt.legend` →  `axes[...].plot`, `axes[...].legend`)
*   Some functions have new name `set_...` (`xticks`, `ylabel` → `set_xticks`,  `set_ylabel`)




```python

fig, axes = plt.subplots(ncols=1, nrows=2, figsize=(10,10),
                         sharex=True #Easy to set sharing of axises
                         )


#Call axes[0] instead of plt
axes[0].plot(years, total_population, color=hse_color,
         marker='o',
         ms=5,
         label='Total Population'
         )
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Population')
axes[0].set_title('Europe Population')
axes[0].set_yticks([ 550000000,     600000000, 650000000, 700000000, 750000000],
           ['550,000,000', '600,000,000', '650,000,000', '700,000,000', '750,000,000']
           )
axes[0].grid(True)
axes[0].plot(forecast_years, forescat_total_population, color=social_science_color,
         marker='s',
         ms=5,
         label='Forecast of Total Population'
         )
axes[0].legend()


#Call axes[1] instead of plt
axes[1].plot(years, migrants, color=hse_color,
         marker='o',
         ms=5,
         label='Migrants'
         )
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Migrants')
axes[1].set_title('Europe Migrants')
axes[1].ticklabel_format(style='plain')


axes[1].plot(forecast_years, forescat_migrants, color=social_science_color,
         marker='s',
         ms=5,
         label='Forecast of Migrants'
         )
axes[1].grid(True)
axes[1].legend()
#### End of copy paste ####


plt.show()

```

## Statistics

##Measures of Central Tendency

Measures of Central Tendency:



*   Mean (Arithmetic)
*   Median
*   Mode



### Mean `.mean()`

$$
\bar {x} =  \frac {1}{n} \sum _{i=1}^{n}x_{i}
$$

```python
df['Age'].mean(), df['Fare'].mean()
```

Problem with mean value: outliers (data points that are significantly different)

```python
#Example:

#Sample 99 times 1 and 1 time 1000
sample = np.ones(100)
sample[0] = 1000

np.mean(sample)
```

### Median `.median()`

Median is the middle score for a set of data that has been arranged in order of magnitude

Example for even number of observations:

Sample: 1 5 3 6

Sort values: 1 3 5 6

The median will be (3 + 5) / 2 = 4





Example for odd number of observations:

Sample: 1 5 3 6 2

Sort values: 1 2 3 5 6

The median will be 3 (the middle value)

```python
df['Age'].median(), df['Fare'].median()
```

Let's check outliers:

```python
np.median(sample)
```
